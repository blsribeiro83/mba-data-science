# Text Mining com Python I

Texto bruto ->  Limpeza (cleaning) -> Representação numérica -> Análise/Modelo

Pré-processamento

1- **Lowercasing**: padronizar o texto. "Eu amo DS!" = ["eu", "amo", "ds", "!"]

2- **Remoção de pontuação**: eliminar sinais de pontuação. "Eu amo DS!" = ["eu","amo","ds"]

3- **Tokenização**: dividir um texto em unidades menores. "Eu amo DS!" = ["Eu","amo","DS","!"]

4- **Stopwords**: palavras comuns que não agregam muito significado ao conteúdo. Ex: "a","o","de","que","em","e"

5- **Stemming**: reduzir palavras à sua raiz. "Eu amor DS!" = Eu am DS

6- **Lemmatization**: reduz a palavra à sua forma canônica. "Amasse" = "amar"

**Text mining** = processo de extrair **informação útil** de dados em texto livre: e-mail, tweets, notícias etc

`Texto bruto → Limpeza (cleaning) → Representação numérica → Análise/Modelo`

`contar → identificar ruído → tokenizar → limpar → recontar`

### Vocabulário
- **Corpus** - conjunto completo de textos que vamos analisar
- **Document** - cada unidade de texto do corpus
- **Token** - cada pedaço mínimo do texto, em geral uma palavra
- **Vocabulary** - o conjunto de todos os tokens distintos do corpus
- **Feature** - cada coluna numérica que representa o texto (ex: a contagem de uma palavra)
