```python
# =============================================================================
# AS 7 CAMADAS DO MODELO OSI
# =============================================================================
#
# O modelo OSI (Open Systems Interconnection) organiza a comunicação em rede
# em 7 camadas abstratas. Cada camada serve à de cima e é servida pela de baixo.
#
# ┌─────┬──────────────────┬──────────────────────────────────────────────────┐
# │ Nº  │ Camada           │ O que faz / exemplos                             │
# ├─────┼──────────────────┼──────────────────────────────────────────────────┤
# │  7  │ Aplicação        │ Interface com o usuário/software                 │
# │     │ (Application)    │ HTTP, HTTPS, FTP, DNS, SMTP                      │
# ├─────┼──────────────────┼──────────────────────────────────────────────────┤
# │  6  │ Apresentação     │ Codificação, compressão, criptografia            │
# │     │ (Presentation)   │ TLS/SSL, UTF-8, JPEG, JSON                       │
# ├─────┼──────────────────┼──────────────────────────────────────────────────┤
# │  5  │ Sessão           │ Abre, mantém e encerra sessões entre processos   │
# │     │ (Session)        │ Cookies de sessão, autenticação OAuth            │
# ├─────┼──────────────────┼──────────────────────────────────────────────────┤
# │  4  │ Transporte       │ Entrega confiável, controle de fluxo e erros     │
# │     │ (Transport)      │ TCP (confiável), UDP (rápido, sem garantia)      │
# ├─────┼──────────────────┼──────────────────────────────────────────────────┤
# │  3  │ Rede             │ Endereçamento lógico e roteamento entre redes    │
# │     │ (Network)        │ IP, ICMP, roteadores                             │
# ├─────┼──────────────────┼──────────────────────────────────────────────────┤
# │  2  │ Enlace de Dados  │ Transferência entre nós adjacentes, detecção     │
# │     │ (Data Link)      │ de erros no meio físico. MAC, Ethernet, Wi-Fi    │
# ├─────┼──────────────────┼──────────────────────────────────────────────────┤
# │  1  │ Física           │ Bits no meio físico: cabos, sinais, voltagem     │
# │     │ (Physical)       │ RJ45, fibra óptica, frequências de rádio         │
# └─────┴──────────────────┴──────────────────────────────────────────────────┘
#
# ONDE O WEB SCRAPING OPERA:
#   requests.get(url)  →  Camada 7 (HTTP)
#   HTTPS/TLS          →  Camada 6 (criptografia)
#   TCP                →  Camada 4 (garante entrega do HTML)
#   IP                 →  Camada 3 (roteia até o servidor)
#
# MNEMÔNICO (do topo para a base):
#   "All People Seem To Need Data Processing"
#    A        P       S    T   N    D       P
# =============================================================================
```

```python
# HTML - HyperText Markup Language
# Exemplo de uma tag simples:
#<html>
#    <head>
#        <title>A simple webpage</title>
#    </head>
#    <body>
#        <!-- This comment is not displayed in the browser -->
#        <h1>Hello world!</h1>
#    </body>
#</html>
# Each element starting with < and ending with > is called a tag

# ATRIBUTOS: informações extras dentro de uma tag
# <tag atributo="valor">conteúdo</tag>
# Exemplos:
# <p id="preco" class="price_color">£51.77</p>
#      ^id único   ^classe CSS        ^texto (conteúdo)
#
# No BeautifulSoup:
#   tag.get_text()        → extrai o conteúdo ("£51.77")
#   tag["class"]          → acessa o atributo class
#   tag["title"]          → acessa o atributo title
#   tag["value"]          → acessa o value de um <input>
```

```python
# ## CSS - Cascading Style Sheets
# Define the appearance of HTML elements on a web page
# h1 {
#   font-size: 20px;
#   color: green;
# }
# h1 is called selector
#
# SELETORES CSS — usados também no BeautifulSoup via .select()
# "article.product_pod h3 a"  → tag article com classe product_pod,
#                                 descendente h3, descendente a
```

```python
# JavaScript
# Client-side code = any code that is sent over by a web server
# but actually executed by the client's browser.
# JavaScript is the only client-side programming language available,
# embedded into HTML using <script> tag:
#<script>
#   alert("Hello, world!");
#</script>
# Or referenced in a separate file using src attribute:
#<script src="someprogram.js"></script>
#
# IMPORTANTE PARA SCRAPING: quando o conteúdo de uma página é gerado
# por JavaScript (renderizado no browser, não no servidor), o HTML
# devolvido pelo requests.get() vem VAZIO — o dado não está lá.
# Nesses casos requests + BeautifulSoup não funcionam.
# Solução: Selenium (controla o browser) ou inspecionar a API
# que o JavaScript chama nos bastidores (aba Network do DevTools).

#JSON - JavaScript Object Notation = is a text format that contains human-readable data
```

```python
# =============================================================================
# CRAWLER vs. WEB SCRAPER
# =============================================================================
#
# Os termos são usados como sinônimos no mercado, mas têm papéis distintos.
#
# SCRAPER — resolve um problema de EXTRAÇÃO
#   Dado um HTML já em mãos, identifica e coleta os dados de interesse.
#   Não decide para onde ir — alguém já forneceu a URL.
#
# CRAWLER — resolve um problema de NAVEGAÇÃO
#   Descobre automaticamente novas URLs a partir de links encontrados
#   nas páginas, decide quais seguir e percorre o site de forma autônoma.
#   Não necessariamente extrai dados — pode só mapear território.
#
# Na prática quase todo projeto real combina os dois:
#   Crawler → descobre URLs → Scraper → extrai dados de cada URL
#
# -----------------------------------------------------------------------------
# QUANDO USAR CADA UM (Mitchell, Cap. 3)
#
# Use um SCRAPER quando:
#   - O conjunto de páginas é fixo e conhecido antecipadamente
#   - O projeto é pontual (one-shot): entrega um CSV e encerra
#   - A estrutura do site é estável e previsível
#   - Exemplos: coletar preços de uma lista de produtos,
#     monitorar cotações de fontes específicas (Projeto 3 deste notebook)
#
# Use um CRAWLER quando:
#   - Você não sabe quantas páginas existem antes de começar
#   - O projeto precisa descobrir conteúdo novo continuamente
#   - O escopo cresce dinamicamente (links levam a mais links)
#   - Exemplos: indexar um catálogo inteiro, monitorar novas publicações,
#     mapear menções de uma marca em múltiplos domínios
#
# -----------------------------------------------------------------------------
# CLASSIFICAÇÃO DE PROJETOS (Mitchell, Cap. 3)
#
# Targeted (direcionado) → URLs conhecidas, estrutura previsível
#   → scraper é suficiente; código mais simples e barato de manter
#
# Broad (amplo) → domínios desconhecidos, estrutura variável
#   → exige crawler; requer arquitetura mais robusta e monitoramento
#
# One-shot → roda uma vez, entrega os dados, código vai para o lixo
#   → pode ser rápido e sujo; o que importa é o dado final
#
# Recorrente → roda periodicamente, atualiza dados, detecta mudanças
#   → exige código robusto, tratamento de erros, infraestrutura de
#     monitoramento para detectar falhas silenciosas
#
# -----------------------------------------------------------------------------
# PONTOS DE ATENÇÃO — TÉCNICOS
#
# Scrapers falham quando:
#   - O site muda o layout (classes CSS, estrutura HTML)
#   - O conteúdo é gerado por JavaScript (requests retorna HTML vazio)
#   - O servidor bloqueia por User-Agent, rate limit ou CAPTCHA
#
# Crawlers têm riscos adicionais:
#   - Escopo pode explodir: um link leva a mil, que levam a um milhão
#   - Loops infinitos se o site tiver links circulares
#   - Carga acumulada no servidor cresce com o volume de requisições
#   - Solução: time.sleep(), limitar profundidade, controlar URLs visitadas
#
# -----------------------------------------------------------------------------
# ÉTICA E LEGALIDADE (Mitchell, Cap. 2 e 3)
#
# Scraper e Crawler compartilham as mesmas obrigações éticas,
# mas o crawler amplia o risco proporcionalmente ao seu alcance.
#
# Antes de qualquer projeto:
#   1. Verifique robots.txt → site.com/robots.txt
#      Disallow indica onde você não deve ir.
#      O que não está bloqueado é consentimento implícito.
#
#   2. Leia os Termos de Serviço (ToS)
#      Muitos proíbem scraping explicitamente.
#      Violar ToS + causar dano + agir intencionalmente
#      = exposição a Trespass to Chattels (eBay v. Bidder's Edge)
#
#   3. Prefira a API oficial quando existir
#      É mais estável, mais ética e menos propensa a bloqueios.
#      O Projeto 3 deste notebook faz exatamente isso com a Wise.
#
#   4. Não colete dados pessoais em escala
#      Pete Warden coletou 200 milhões de perfis públicos do Facebook.
#      Resultado: cease and desist. No Brasil: exposição à LGPD
#      (Art. 7º §3º — dados públicos ainda exigem finalidade e boa-fé).
#
#   5. Throttle suas requisições — sempre
#      Um crawler sem time.sleep() pode derrubar servidores pequenos.
#      Isso configura dano real → critério de Trespass to Chattels.
#      Regra prática: rode de madrugada, devagar, em round-robin
#      quando o alvo forem múltiplos domínios.
#
#   6. Não publique conteúdo criativo scrapeado
#      Armazenar para análise privada → permitido
#      Publicar texto, imagens, obras de terceiros → violação de copyright
#      Publicar estatísticas e análises derivadas → permitido
#
#   7. Se encontrar vulnerabilidade: reporte à empresa, não à mídia
#      Auernheimer coletou 114 mil emails de uma URL pública da AT&T
#      e entregou à Gawker. Resultado: 3 anos de processo federal.
#
# Resumo ético de Mitchell em uma frase:
#   "Don't be a jerk."
# =============================================================================
```

```python
# =============================================================================
# CAPÍTULO 3 — APLICAÇÕES DE WEB SCRAPING (Mitchell)
# =============================================================================
#
# Capítulo estratégico, não técnico. Tese central: scraping sem planejamento
# é trabalho jogado fora. Antes do código, classifique o projeto.
#
# -----------------------------------------------------------------------------
# 1. CLASSIFICAÇÃO DE PROJETOS — três perguntas obrigatórias
# -----------------------------------------------------------------------------
#
# ESCOPO: targeted ou broad?
#   Targeted → URLs conhecidas, estrutura previsível, scraper é suficiente
#   Broad    → domínios desconhecidos, estrutura variável, exige crawler
#              e capacidade de lidar com layouts completamente diferentes
#
# FREQUÊNCIA: one-shot ou recorrente?
#   One-shot    → o produto é o dado, não o código
#                 pode ser rápido e sujo; entrega CSV, código vai para o lixo
#   Recorrente  → o código vira ativo de negócio
#                 precisa ser robusto, manutenível, e monitorado
#                 falha silenciosa é o pior tipo de falha em scraping
#
# PROFUNDIDADE: dado bruto ou insumo para análise?
#   Coleta simples  → deposita como encontra, limpeza mínima
#   Análise avançada → colete mais do que o necessário agora;
#                      voltar ao site depois é mais caro
#
# -----------------------------------------------------------------------------
# 2. DOMÍNIOS DE APLICAÇÃO
# -----------------------------------------------------------------------------
#
# E-COMMERCE
#   Caso de uso mais comum. Vai além de preço:
#   - Variações de produto (tamanho, cor) com disponibilidade e custo
#   - SKU: identificador único por variação, frequentemente escondido
#     no HTML ou em APIs JavaScript — vale procurar
#   - Estoque: rastrear quando produto sai e volta ao estoque
#     revela demanda de mercado
#   - Histórico de promoções: preço original + preço de venda separados
#     permitem mapear estratégia de desconto do concorrente
#   - Reviews e ratings: podem ser analisados, nunca republicados verbatim
#
# MARKETING
#   - Monitoramento de marca e sentimento público
#   - Detecção de uso indevido de trademark por concorrentes
#   - Identificação de micro e nano-influenciadores por frequência
#     de postagem sobre tópicos relevantes
#   ⚠️  Redes sociais: poucos sites, dados bem formatados, mas alto risco
#       legal. Mitchell recomenda explicitamente consultar advogado
#       antes de scrapar Facebook e LinkedIn.
#
# PESQUISA ACADÊMICA
#   - NIH usou scraping para estimar prevalência de HIV em presídios
#     → passou por aprovação ética formal antes de rodar
#   - Guardian: análise de retórica em comentários sobre obesidade
#   - Banco do Japão: scraping para ajustar índices de preços
#     com "dados alternativos" fora do fluxo tradicional (PIB, relatórios)
#   ⚠️  Em pesquisa médica ou com dados sensíveis:
#       análise ética precede o código — sempre
#
# CONSTRUÇÃO DE PRODUTO
#   Quando você precisa de base de dados para MVP e não há API disponível
#   ou ela é cara demais:
#   - Sites de viagem: pontos turísticos (não apenas coordenadas geográficas)
#   - Datasets de treinamento para modelos de ML
#   - Listas de fornecedores, contatos, vagas em nichos específicos
#   Scraping como forma de reduzir custo de aquisição de dados na fase inicial
#
# VIAGEM
#   Hotéis, aéreas e locadoras: pouca diferenciação, muita concorrência,
#   flutuações constantes de preço → terreno ideal para scraping.
#   Kayak e Trivago começaram exatamente assim.
#   Defesas anti-scraping moderadas: se um site bloquear,
#   há outro com os mesmos dados.
#   Mitchell recomenda o setor como projeto de aprendizado:
#   volume alto, variação temporal, desafios de engenharia reais.
#
# VENDAS E RECRUTAMENTO
#   - Geração de leads e listas de contato segmentadas
#   - Yelp: prospecção de negócios locais com filtros detalhados
#   - LinkedIn: políticas anti-scraping agressivas
#     → plugins de browser (Instant Data Scraper, Dux-Soup) são usados
#     para raspar páginas visitadas manualmente, contornando bloqueios
#
# SERP SCRAPING
#   Raspar a página de resultados do Google em vez de seguir os links.
#   Vantagem: formato consistente, sem lidar com layouts variados.
#   Casos de uso:
#   - Histórico de standings esportivos
#   - Monitoramento de posicionamento de marca nos resultados
#   - Rastreamento de anúncios de concorrentes
#   O Google tem verticais separadas (Maps, Images, Shopping, Flights, News)
#   — cada uma é na prática um buscador distinto para seu tipo de conteúdo
#
# -----------------------------------------------------------------------------
# 3. ÉTICA — transversal a todos os domínios
# -----------------------------------------------------------------------------
#
# Mitchell não isola ética em uma seção — ele a embute em cada domínio.
# Os casos de maior risco:
#
# Dados pessoais em escala
#   → exigem análise ética formal antes do código
#   → no Brasil: conformidade com LGPD (Art. 7º §3º)
#     dados públicos ainda exigem finalidade e boa-fé
#
# Redes sociais
#   → proteção por tecnologia E por advogados (palavras de Mitchell)
#   → Facebook e LinkedIn: zona de alto risco, consulte jurídico
#
# Conteúdo criativo
#   → pode ser coletado para análise privada
#   → nunca para republicação (copyright automático — Cap. 2)
#
# Projetos recorrentes em sites pequenos
#   → scraper mal throttled onera servidor continuamente
#   → dano real contínuo = critério de Trespass to Chattels (Cap. 2)
#
# -----------------------------------------------------------------------------
# CONEXÃO COM ESTE NOTEBOOK
#
# O Projeto 3 (comparação de cotação do dólar) é exatamente o caso de uso
# de construção de produto na categoria finanças/viagem.
# O design já reflete as lições do capítulo:
#   - Prefere API quando existe            → Wise via JSON
#   - Usa BeautifulSoup só quando precisa  → DolarHoje (campo <input>)
#   - Usa pd.read_html para tabelas        → x-rates
#   - Encapsula em try/except              → sites reais mudam e falham
# =============================================================================
```

```python
# Criando uma função para obter o título ou enviar mensagem de erro
# Abaixo usou-se dois try para testar a conexão de rede e para testar a estrutura do HTML

from urllib.request import urlopen
from urllib.error import HTTPError
from bs4 import BeautifulSoup

def getTitle (url):
    try:
        html = urlopen(url)                                    #tenta abrir o site
    except HTTPError as e:
        return None                                            #se o site retornar erro, para aqui
    try:
        bs = BeautifulSoup(html.read(),"html.parser")          #transforma HTML em árvore
        title = bs.body.h1                                     #tenta acessar o h1
    except AttributeError as e:
        return None
    return title
```

```python
#Encontrar uma tag específica
# usamos bs.get_text()    para retornar apenas o texto, sem as tags
titulo = soup.find("h1")
print ("Título da página: ",titulo.get_text(strip=True))
```

```python
#Encontrar todas as tags
cartoes = soup.find_all("div",class_="bg-white")
print ("Cidades encontradas:", len(cartoes))
```

```python
# Para cada cartão, extraímos cidade + os três valores (na ordem das spans)
dados = []
for cartao in cartoes:
    cidade = cartao.find("h2").get_text(strip=True)
    valores = cartao.find_all("span",class_="font-bold")
    temperatura = valores[0].get_text(strip=True)
    clima = valores[1].get_text(strip=True)
    umidade = valores[2].get_text(strip=True)
    dados.append({
        "cidade":cidade,
        "temperatura":temperatura,
        "clima":clima,"umidade":umidade
    })
    print (f"{cidade:15} | {temperatura:6} | {clima:22} | umidade {umidade}")
dados_final = pd.DataFrame(dados)
dados_final
```

```python
livros = soup.find_all("article",class_="product_pod")
print("Livros nesta página: ",len(livros))

dados = []
for livro in livros:
    titulo = livro.find("h3").find("a")["title"]
    preco = livro.find("p",class_="price_color").get_text(strip=True)
    avaliacao = livro.find("p",class_="star-rating")["class"][1]
    estoque = livro.find("p",class_="instock").get_text(strip=True)
    dados.append({"titulo":titulo,"preco":preco,"avaliacao":avaliacao,"estoque":estoque})
    print (f"{avaliacao:6}  |  {preco:8}  |  {titulo}")

dadoslivros = pd.DataFrame(dados)
dadoslivros
```

```python
# Escrevendo um crawler para percorrer todas as páginas
# padrão catalogue/page-N.html
# usar time.sleep para não sobrecarregar o servidor
from bs4 import BeautifulSoup
import pandas as pd
import requests
import time
todos = []
for pagina in range (1,2): #páginas 1, 2 e 3
    url = f"https://books.toscrape.com/catalogue/page-{pagina}.html"
    r = requests.get(url)
    if r.status_code != 200:
        break
    s = BeautifulSoup(r.text,"lxml")
    genero = s.find("div", class_="page-header").find("h1").get_text(strip=True)
    for livro in s.find_all("article",class_="product_pod"):
        todos.append({
            "titulo":livro.find("h3").find("a")["title"],
            "preco": livro.find("p", class_="price_color").get_text(strip=True),
            "nota":livro.find("p",class_="star-rating")["class"][1],
            "genero":genero,
        })
        time.sleep(1)
todos = pd.DataFrame(todos)
todos
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from io import StringIO
```

```python
# Cabeçalho de navegador: alguns sites respondem melhor quando nos identificamos
HEAD = {"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:152.0) Gecko/20100101 Firefox/152.0"}

resultados = []

def para_float(text):
    
    return float(str(text).replace(".",",").replace(",","."))
```

```python
# Fonte 1 - Wise (provedor real) via API
try:
    url = "https://wise.com/rates/live?source=USD&target=BRL"
    valor = requests.get(url,headers = HEAD, timeout =15).json()["value"]
    valor = float(valor)


    print(f"Wise - taxa USD -> BRL: R$ {valor:.4f}")
    resultados.append({"fonte": "Wise", "tipo": "comercial", "preco_venda_usd": valor})

except Exception as e:
    print ("Falha ao coletar da Wise:", e)
```

```python
# Fonte 2 - Site real com BeautifulSoup - DolarHoje - Dólar turismo
try:
    url = "https://dolarhoje.com/dolar-turismo/"
    html = requests.get(url, headers = HEAD, timeout = 15).text
    soup = BeautifulSoup(html,"lxml")

    campo = soup.find("input",id="nacional")
    valor = para_float(campo["value"])

    print(f"DolarHoje Turismo: R$ {valor:.4f}")

    resultados.append({"fonte":"DolarHoje","tipo":"turismo","preco_venda_us":valor})

except Exception as e:
    print(f"Falha ao coletar DolarHoje:",e)
```
