```python
# importações
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
URL = "https://books.toscrape.com/"
res = requests.get(URL)
bs = BeautifulSoup(res.text,"lxml")
```

```python
bs.find("h1").get_text(strip=True)
```

```python
livros = bs.find_all("article",class_="product_pod")
dados = []
for livro in livros:
    titulo = livro.find("h3").find("a")["title"]
    preco = livro.find("p",class_="price_color").get_text(strip=True)
    estoque = livro.find("p",class_="instock availability").get_text(strip=True)
    avaliacao = livro.find("p",class_="star-rating")["class"][1]
    dados.append({"titulo":titulo,"preco":preco,"estoque":estoque,"avaliacao":avaliacao})
livros_df = pd.DataFrame(dados)
livros_df
```

```python
URL = "https://quotes.toscrape.com/"
res = requests.get(URL)
bs = BeautifulSoup(res.text,"lxml")

quotes = bs.find_all("div",class_="quote")
#tags = bs.find_all("div",class_="tags")
todos = []
for quote in quotes:
    citacao = quote.find("span",class_="text").get_text(strip=True)
    autor = quote.find("small",class_="author").get_text(strip=True)
    
    dados_tag = []
    for tag in quote.find_all("a",class_="tag"):
        dados_tag.append(tag.get_text(strip=True))

    todos.append({"citacao":citacao,"autor":autor,"tag":dados_tag})

citacao = pd.DataFrame(todos)
citacao
```

```python
URL = "https://webscraper.io/test-sites/pagination"
import re

res = requests.get(URL)
bs = BeautifulSoup(res.text,"lxml")

cars = bs.find_all("div",class_="card")

todos = []

for car in cars:
    carro = car.find("h3").find("a").get_text(strip=True)
    modelo = car.find("p",class_="description").get_text(strip=True)
    dados = car.find_all("p",class_="card-text")
    ano = dados[0].get_text(strip=True)
    ano = ano.replace("Year:","").strip()
    pais = dados[1].get_text(strip=True)
    pais = pais.replace("Country of origin:","").strip()
    milhagem = dados[2].get_text(strip=True)
    milhagem = milhagem.replace("Mileage:","").strip()
    todos.append({"carro":carro,"modelo":modelo,"ano":ano,"pais":pais,"milhagem":milhagem})
    
todos = pd.DataFrame(todos)
todos
```
