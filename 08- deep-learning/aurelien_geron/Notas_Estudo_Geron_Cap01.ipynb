## Machine Learning - Chapter 1

Machine learning - the science and ard of programming computersa so they can **learn from data**

```python
# Experience E, task T, performance measure P
# Its performance on T, measured by P, improves with experience E
# The examples that the system uses to learn are called training set
# Each training sample is called training instance (or sample)
# The part of a machine learning that learns and makes predictions is called a model
```

```python
# Critérios de classificação de Machine Learning
# Se são supervisionados durante o treinamento
# Se aprendem ou não de forma incremental durante seu uso
# Se funcionam comparando novas informações com informações existentes ou se detectam padrões nos dados de treinamento e constroem um modelo preditivo
```

```python
# Aprendizado supervisionado: o conjunto de treinamento para o algoritmo inclui a solução desejada, chamada label
# Poussi class (o valor esperado no treinamento) e features (informações que ajudam a treinar o modelo, também chamados predictors or attributes)
# Exemplo típico: classificação - utiliza labels
# Regression é o típico modelo de ML relacionado a predizer um valor - utiliza targets
```

```python
# Aprendizado não-supervisionado: o conjunto de treinamento de dados não possui label; o sistema tenta aprender sem um professor
# Dimensionality reduction - algoritmo cujo objetivo é simplificar dados sem perder muita informação
# Boa prática: usar um algoritmo de DR para otimizar o número de dimensões antes de alimentar em um modelo de ML
# Anomaly detection - algoritmo utilizado para detectar eventos fraudulentos ou errôneos
# Novelty detection - algoritmo que tem por objetivo mirar em novas istâncias que parecem diferentes das instâncias existentes
# Association rule learning - algortimo que mergulha em grandes volumes de dados e descobre relações entre atributos
```

```python
# Aprendizado semissupervisionado - algoritmos que trabalham com dados parcialmente rotulados
# Ex: sistema do Google Photos, que identifica uma pessoa a partir de um rótulo e dá o mesmo rótulo a essa pessoa em outras fotos
```
