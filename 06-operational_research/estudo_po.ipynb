```python
problema = pulp.LpProblem("corte_de_bobinas",pulp.LpMaximize)

x1 = pulp.LpVariable("x1_maquina_antiga",lowBound=0)
x2 = pulp.LpVariable("x2_maquina_nova",lowBound=0)

problema += 0.3 * x1 + 0.5 * x2
problema += x1 <= 4000
problema += x2 <= 6000
problema += (3/1000) * x1 + (2/1000) * x2 <= 18

problema.solve(pulp.PULP_CBC_CMD(msg=0))

print (f"x1 = {x1.value()} metros/dia")
print (f"x2 = {x2.value()} metros/dia")
print (f"Lucro máximo = R$ {pulp.value(problema.objective):.2f}")
```

```python
import pulp

# definição do problema
problema = pulp.LpProblem("caixas",pulp.LpMaximize)

# variáveis
# cl = caixa leve
# cm = caixa mista

cl = pulp.LpVariable("caixa_leve",lowBound=0,cat="integer")
cm = pulp.LpVariable("caixa_mista",lowBound=0,cat="integer")

# função objetivo e funções de restrição

problema += 6*cl + 7 * cm
problema += cl <= 10
problema += cm <= 7
problema += cm >= 3
problema += 3 * cl + cm >= 20

problema.solve(pulp.PULP_CBC_CMD(msg=True))

print(cl.value())
print(cm.value())
print(pulp.LpStatus[problema.status])
```

```python
# problema - nova ala de hospital

import pulp

# definição do problema

problema = pulp.LpProblem("leitos_hospital",pulp.LpMaximize)

# definição variáveis
# q1 = quarto com 1 leito
# q2 = quarto com dois leitos
# q3 = quarto com três leitos

q1 = pulp.LpVariable("quarto_um_leito",lowBound=0,cat="integer")
q2 = pulp.LpVariable("quarto_dois_leitos",lowBound=0,cat="integer")
q3 = pulp.LpVariable("quarto_tres_leitos",lowBound=0,cat="integer")

# definição função-objetivo

problema += 7500 * q1 + 6000 * q2 + 4500 * q3 #problema.objective

# restrições

problema += 10*q1+15*q2+18*q3 <= 1200
problema += q1 + q2 + q3 <= 70
problema += q1 + 2*q2 + 3*q3 >= 120
problema += q1 <= 30

problema.solve(pulp.PULP_CBC_CMD(msg=0))

print (f"{q1.value()} quartos de um leito")
print(f"{q2.value()} quartos de dois leito")
print (f"{q3.value()} quartos de três leitos")
print (f"Renda total: {pulp.value(problema.objective)}")
print(pulp.LpStatus[problema.status])
```

```python
import pulp

# descrição do problema

problema = pulp.LpProblem("racao",pulp.LpMinimize)

# variáveis
x1 = pulp.LpVariable("cevada",lowBound=0)
x2 = pulp.LpVariable("aveia",lowBound=0)
x3 = pulp.LpVariable("soja",lowBound=0)
x4 = pulp.LpVariable("milho",lowBound=0)

# equação objetivo
problema += 30*x1 + 48*x2 + 44*x3 + 56*x4

#equações de restrição
problema += x1 + x2 + x3 + x4 == 10000 # quantidade de ração
problema += 6.9/100 * x1 + 8.5/100 * x2 + 9/100 * x3 + 27.1/100 * x4 == 0.15 * 10000, "Restricao_Proteina" # proteina
problema += 6/100 * x1 + 11/100 * x2 + 11/100 * x3 + 14/100 * x3 >= 0.08 * 10000, "Restricao_Fibra" # fibra
problema += 1760*x1 + 1700*x2 + 1056*x3 + 1400*x4 >= 1100 * 10000, "Restricao_Calorias_Minima" #calorias
problema += 1760*x1 + 1700*x2 + 1056*x3 + 1400*x4 <= 2250 * 10000, "Restricao_Calorias_Maxima"
problema += x2 >= 0.20 * 10000, "Restricao_Percentual_Milho"
problema += x3 <= 0.12 * 10000, "Restricao_Percentual_Soja"

# solver
# problema.solve(pulp.PULP_CBC_CMD(msg=0))
problema.solve()

print(f"Cevada: {x1.value()} kg")
print (f"Aveia: {x2.value()} kg")
print (f"Soja: {x3.value()} kg")
print (f"Milho: {x4.value()} kg")
print (f"Custo total: {pulp.value(problema.objective)}")
print (x1.value() + x2.value() + x3.value() + x4.value())
print(pulp.LpStatus[problema.status])
```

```python
# Instalar o PuLP
# Executar o seguinte comando diretamente no console: pip install pulp

# Importar o PuLP
from pulp import *

# Definir o problema
prob = LpProblem("Corte_Bobinas", LpMaximize)

# Definir as variáveis de decisão
x = LpVariable("x", lowBound=0, cat='Continuous')  # Produção na máquina antiga
y = LpVariable("y", lowBound=0, cat='Continuous')  # Produção na máquina moderna

# Definir a função objetivo
prob += 0.3 * x + 0.5 * y, "Lucro_Total"

# Adicionar as restrições
prob += x <= 4000, "Capacidade_Maquina_Antiga"
prob += y <= 6000, "Capacidade_Maquina_Moderna"
prob += 0.003 * x + 0.002 * y <= 18, "Disponibilidade_Mao_de_Obra"

# Resolver o problema
prob.solve()

# Imprimir os resultados
print("Status:", LpStatus[prob.status])
print(f"Produção na Máquina Antiga (x) = {x.varValue:.2f} metros")
print(f"Produção na Máquina Moderna (y) = {y.varValue:.2f} metros")
print(f"Lucro Total = R$ {value(prob.objective):.2f}")
```

```python
# Instalar o PuLP
# Executar o seguinte comando diretamente no console: pip install pulp

# Importar a biblioteca PuLP
from pulp import *

# Definir o problema como de maximização
problema = LpProblem("Maximizar_Receita_Mensal", LpMaximize)

# Definir as variáveis de decisão: q1, q2 e q3 (número de quartos individuais, duplos e triplos)
q1 = LpVariable("q1", lowBound=0, cat="Integer")  # Quartos individuais (inteiros e ≥ 0)
q2 = LpVariable("q2", lowBound=0, cat="Integer")  # Quartos duplos (inteiros e ≥ 0)
q3 = LpVariable("q3", lowBound=0, cat="Integer")  # Quartos triplos (inteiros e ≥ 0)

# Definir a função objetivo: R = 7500*q1 + 6000*q2 + 4500*q3
problema += 7500*q1 + 6000*q2 + 4500*q3, "Receita_Total"

# Adicionar as restrições
problema += 10*q1 + 15*q2 + 18*q3 <= 1200, "Restricao_de_Area"         # Restrição de área total
problema += q1 + q2 + q3 <= 70, "Restricao_de_Capacidade"              # Restrição de número total de quartos
problema += q1 <= 30, "Restricao_de_Quartos_Individuais"              # Limite máximo de quartos individuais
problema += q1 + 2*q2 + 3*q3 >= 120, "Restricao_de_Ocupacao_Minima"   # Restrição de ocupação mínima

# Resolver o problema
problema.solve()

# Exibir o status da solução
print("Status da Solução:", LpStatus[problema.status])

# Exibir os valores ótimos das variáveis
print("Quantidade ótima de quartos individuais (q1):", q1.varValue)
print("Quantidade ótima de quartos duplos (q2):", q2.varValue)
print("Quantidade ótima de quartos triplos (q3):", q3.varValue)

# Exibir o valor máximo da função objetivo
print("Receita mensal máxima (R): R$", value(problema.objective))
```

```python
import pulp

problema = pulp.LpProblem("guardanapos", pulp.LpMinimize)

# demanda por dia (dado do enunciado)
d = {1: 110, 2: 210, 3: 190, 4: 120, 5: 100}

# --- variáveis ---
# compra de guardanapo novo para o dia i
x1 = pulp.LpVariable("novo_dia1", lowBound=0)
x2 = pulp.LpVariable("novo_dia2", lowBound=0)
x3 = pulp.LpVariable("novo_dia3", lowBound=0)
x4 = pulp.LpVariable("novo_dia4", lowBound=0)
x5 = pulp.LpVariable("novo_dia5", lowBound=0)

# quantidade enviada para lavanderia 24h no dia i (retorna disponível no dia i+2)
y1 = pulp.LpVariable("24h_enviado_dia1", lowBound=0)
y2 = pulp.LpVariable("24h_enviado_dia2", lowBound=0)
y3 = pulp.LpVariable("24h_enviado_dia3", lowBound=0)

# quantidade enviada para lavanderia 48h no dia i (retorna disponível no dia i+3)
t1 = pulp.LpVariable("48h_enviado_dia1", lowBound=0)
t2 = pulp.LpVariable("48h_enviado_dia2", lowBound=0)

# guardanapos jogados fora direto (sem lavar) após o uso no dia i
v1 = pulp.LpVariable("lixo_dia1", lowBound=0)
v2 = pulp.LpVariable("lixo_dia2", lowBound=0)
v3 = pulp.LpVariable("lixo_dia3", lowBound=0)
v4 = pulp.LpVariable("lixo_dia4", lowBound=0)
v5 = pulp.LpVariable("lixo_dia5", lowBound=0)

# --- função objetivo: minimizar custo total ---
problema += (
    10 * (x1 + x2 + x3 + x4 + x5)   # guardanapos novos: $10/un
    + 5 * (y1 + y2 + y3)             # serviço 24h: $5/un
    + 3 * (t1 + t2)                  # serviço 48h: $3/un
), "Custo_Total"

# --- restrições de DEMANDA: como cada dia é suprido ---
problema += x1 == d[1], "Demanda_Dia1"            # só novo (sem retorno ainda)
problema += x2 == d[2], "Demanda_Dia2"            # só novo (sem retorno ainda)
problema += x3 + y1 == d[3], "Demanda_Dia3"       # novo + 24h enviado dia1
problema += x4 + y2 + t1 == d[4], "Demanda_Dia4"  # novo + 24h enviado dia2 + 48h enviado dia1
problema += x5 + y3 + t2 == d[5], "Demanda_Dia5"  # novo + 24h enviado dia3 + 48h enviado dia2

# --- restrições de SUJEIRA: destino dos guardanapos usados em cada dia ---
problema += y1 + t1 + v1 == d[1], "Sujeira_Dia1"  # usados no dia1: vão p/ 24h, 48h ou lixo
problema += y2 + t2 + v2 == d[2], "Sujeira_Dia2"  # usados no dia2
problema += y3 + v3 == d[3], "Sujeira_Dia3"       # usados no dia3 (48h não cabe mais no horizonte)
problema += v4 == d[4], "Sujeira_Dia4"            # usados no dia4: sem tempo de lavar, direto lixo
problema += v5 == d[5], "Sujeira_Dia5"            # usados no dia5: convenção acabou, direto lixo

problema.solve(pulp.PULP_CBC_CMD(msg=0))

print("Status:", pulp.LpStatus[problema.status])
print(f"Novos comprados : dia1={x1.value()}, dia2={x2.value()}, dia3={x3.value()}, dia4={x4.value()}, dia5={x5.value()}")
print(f"Enviados 24h    : dia1={y1.value()}, dia2={y2.value()}, dia3={y3.value()}")
print(f"Enviados 48h    : dia1={t1.value()}, dia2={t2.value()}")
print(f"Jogados no lixo : dia1={v1.value()}, dia2={v2.value()}, dia3={v3.value()}, dia4={v4.value()}, dia5={v5.value()}")
print(f"Custo total = $ {pulp.value(problema.objective):.2f}")
```
