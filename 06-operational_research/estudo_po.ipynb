```python
import pulp
```

```python
problema = pulp.LpProblem("corte_de_bobinas",pulp.LpMaximize)

x1 = pulp.LpVariable("x1_maquina_antiga",lowBound=0)
x2 = pulp.LpVariable("x2_maquina_nova",lowBound=0)

problema += 0.3 * x1 + 0.5 * x2
problema += x1 <= 4000
problema += x2 <= 6000
problema += (3/1000) * x1 + (2/1000) * x2 <= 18

problema.solve(pulp.PULP_CBC_CMD(msg=0))

print (f"x1 = {x1.value()} metros/dia")
print (f"x2 = {x2.value()} metros/dia")
print (f"Lucro máximo = R$ {pulp.value(problema.objective):.2f}")
```

```python
import pulp

problema = pulp.LpProblem("programaca_horas",pulp.LpMaximize)

# variáveis de estudo
# x1 = maquina mais antiga
# x2 = maquina mais nova

x1 = pulp.LpVariable("maquina_antiga",lowBound=0)
x2 = pulp.LpVariable("maquina_nova",lowBound = 0)

# equação a maximizar
problema += x1 * 0.3 + x2 * 0.5

# restrições
problema += (3/1000) * x1 + (2/1000) * x2 <= 18
problema += x1 <= 4000
problema += x2 <= 6000

# solver
problema.solve(pulp.PULP_CBC_CMD(msg=0))

print (f"Máquina antiga: {x1.value()} metros/dia")
print (f"Máquina nova: {x2.value()} metros/dia")
```

```python
import pulp

# variáveis
# d1 = disciplina 1
# d2 = disciplina 2

d1 = pulp.LpVariable("disciplina_1",lowBound=0)
d2 = pulp.LpVariable("disciplina_2",lowBound=0)

# problema

problema = pulp.LpProblem("disciplinas",pulp.LpMaximize)

#informações
# d1 = 100/20 = 5 pontos/hora
# d2 = 100/25 = 4 pontos/hora

problema += (3 * d1 * 5 + 5 * d2 * 4) / 8
problema += d1 + d2 <=30
problema += 5 * d1 >= 50
problema += 4 * d2 >= 50
problema += d1 <=20
problema += d2 <= 25

problema.solve(pulp.PULP_CBC_CMD(msg=0))

print (f"d1 = {d1.value()} horas")
print (f"d2 = {d2.value()} horas")
```

```python
import pulp

# definição do problema
problema = pulp.LpProblem("caixas",pulp.LpMaximize)

# variáveis
# cl = caixa leve
# cm = caixa mista

cl = pulp.LpVariable("caixa_leve",lowBound=0,cat="integer")
cm = pulp.LpVariable("caixa_mista",lowBound=0,cat="integer")

# função objetivo e funções de restrição

problema += 6*cl + 7 * cm
problema += cl <= 10
problema += cm <= 7
problema += cm >= 3
problema += 3 * cl + cm >= 20

problema.solve(pulp.PULP_CBC_CMD(msg=0))

print(cl.value())
print(cm.value())
print(pulp.LpStatus[problema.status])
```

```python
# problema - nova ala de hospital

import pulp

# definição do problema

problema = pulp.LpProblem("leitos_hospital",pulp.LpMaximize)

# definição variáveis
# q1 = quarto com 1 leito
# q2 = quarto com dois leitos
# q3 = quarto com três leitos

q1 = pulp.LpVariable("quarto_um_leito",lowBound=0,cat="integer")
q2 = pulp.LpVariable("quarto_dois_leitos",lowBound=0,cat="integer")
q3 = pulp.LpVariable("quarto_tres_leitos",lowBound=0,cat="integer")

# definição função-objetivo

problema += 7500 * q1 + 6000 * q2 + 4500 * q3 #problema.objective

# restrições

problema += 10*q1+15*q2+18*q3 <= 1200
problema += q1 + q2 + q3 <= 70
problema += q1 + 2*q2 + 3*q3 >= 120
problema += q1 <= 30

problema.solve(pulp.PULP_CBC_CMD(msg=0))

print (f"{q1.value()} quartos de um leito")
print(f"{q2.value()} quartos de dois leito")
print (f"{q3.value()} quartos de três leitos")
print (f"Renda total: {pulp.value(problema.objective)}")
print(pulp.LpStatus[problema.status])
```
