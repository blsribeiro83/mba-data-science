## Aula 1 - Decision Tree

```python
# importações
import pandas as pd
```

```python
# leitura do dataset

df = pd.read_csv("https://raw.githubusercontent.com/vqrca/ml-datasets/refs/heads/main/Heart_Disease_Prediction.csv")
```

```python
# exibição das primeiras observações para verificar a estrutura dos dados e as variáveis disponíveis

pd.set_option("display.max_columns",None)
print (df.head())
```

```python
# avaliando o set

df.shape
df.info
```

```python
# Treinando a árvore de decisão

x = df.drop("Heart Disease",axis=1)
y = df["Heart Disease"]
```

```python
x
```

```python
y
```

```python
# Separação de dados em conjuntos de treinamento e teste

from sklearn.model_selection import train_test_split
x_treino,x_teste,y_treino,y_teste = train_test_split(x,y,test_size = 0.2,stratify=y,random_state=42)
```

```python
# Classificação com Decision Tree

from sklearn.tree import DecisionTreeClassifier

modelo_arvore = DecisionTreeClassifier(random_state = 5389)
```

```python
modelo_arvore.fit(x_treino,y_treino)
```

```python
predicoes = modelo_arvore.predict(x_teste)
```

```python
predicoes
```

```python
probabilidades = modelo_arvore.predict_proba(x_teste)
```
