```python
#%% Importando o banco de dados

dados_jogos = pd.read_csv('(4.2) Jogos Copa 22.csv', sep=',')
```

```python
dados_jogos.info()
```

```python
dados_jogos["venceu"] = np.where((dados_jogos['team_home_score']-dados_jogos['team_away_score']>0),\
                                 "Mandante",np.where((dados_jogos['team_home_score']-dados_jogos['team_away_score']<0),\
                                                     "Visitante",np.where((dados_jogos['team_home_score']-dados_jogos['team_away_score']==0),\
                                                                          "Empate","Demais")))
```

```python
dados_jogos
```

```python
# TSubstituir a informação de penaltis para NaN e alterar para tipo float
dados_jogos["pens_home_score"] = dados_jogos['pens_home_score'].replace("False",np.nan)
dados_jogos['pens_away_score']=dados_jogos['pens_away_score'].replace("False",np.nan)
dados_jogos['pens_away_score']=dados_jogos['pens_away_score'].astype("float")
dados_jogos['pens_home_score']=dados_jogos['pens_home_score'].astype("float")
```

```python
dados_jogos
```

```python
# Ajustando a variável com o time vencedor
dados_jogos["vencedor"] = np.where(dados_jogos['pens_home_score']-dados_jogos['pens_home_score']>0,"Mandante",\
                                   np.where(dados_jogos['pens_home_score']-dados_jogos['pens_home_score']<0,\
                                            "Visitante",np.where(dados_jogos['pens_home_score'].isna(),dados_jogos["venceu"],\
                                            "Demais")))
#na última linha, onde não houve pênalti, usa o resultado do DataFrame anterior, com a variável "venceu"
```
