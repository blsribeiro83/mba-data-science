```python
# IMPORTAÇÃO DE BANCOS DE DADOS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# pip install xlrd # pacote dependência que read_excel usa para ler arquivos na extensão .xls
```

```python
# Datasets
dados_filmes = pd.read_csv('(5.2) Filmes Streaming.csv', sep=',')
dados_series = pd.read_csv('(5.3) Séries Streaming.csv', sep=',')
```

```python
dados_filmes.info()
```

```python
dados_series.info()
```

```python
# Selecionando as variáveis de interesse
dados_filmes = dados_filmes.iloc[:,0:12]
```

```python
# Junção de dados
dados_completo = pd.concat([dados_filmes,dados_series],ignore_index=True)
```

```python
# Remoção da variável indesejada
dados_completo.drop(columns=("Unnamed: 0"),inplace=True)
```

```python
# Extrair notas
dados_completo["Ajuste_IMDB"] = dados_completo["IMDb"].str.slice(0,4)
dados_completo["Ajuste_Rotten"]=dados_completo['Rotten Tomatoes'].str.slice(0,3)
```

```python
# Ajustar a string e transformar para float
# A função str.rstrip() remove o caractere ao final da string
# str.rstrip — remove caracteres apenas do lado direito (final) da string
# str.strip — remove caracteres dos dois lados (início e fim)
# str.lstrip — remove só do lado esquerdo (início)
dados_completo['Ajuste_IMDB']=dados_completo['Ajuste_IMDB'].str.rstrip('/').astype("float")
```

```python
dados_completo['Ajuste_Rotten']=dados_completo['Ajuste_Rotten'].str.rstrip('/').astype("float")
```

```python
# Atribuindo labels
muda_numeros = {0:"filme",1:"serie"}
dados_completo = dados_completo.assign(tipo = dados_completo["Type"].map(muda_numeros))
# Aplica o dicionário em cada elemento da coluna Type. Onde encontra 0 coloca 'filme',
# onde encontra 1 coloca 'série'
# Qualquer valor fora do dicionário vira NaN.
```

```python
dados_completo
```

```python
# Agrupando o dataset
descritivas = dados_completo.groupby(['tipo'])
```

```python
descritivas["Ajuste_IMDB"].describe().T
```

```python
descritivas["Ajuste_Rotten"].describe().T
```

```python
# Criar um indicador com os melhores filmes e série, segundo avaliação IMDB e Rotten
# Vamos usar o percentil como referência
melhores_series=dados_completo[dados_completo['tipo']=="serie"]
melhores_filmes=dados_completo[dados_completo['tipo']=="filme"]
```

```python
melhores_series[["Ajuste_IMDB","Ajuste_Rotten"]].quantile(0.95)
```

```python
# gerando dados
melhores_series = melhores_series.assign(Categ_IMDB = pd.qcut\
                                         (melhores_series['Ajuste_IMDB'],q=[0,0.95,1],labels=["menores","maiores"]))
```

```python
melhores_series = melhores_series.assign(Categ_Rotten = \
                                         pd.qcut(melhores_series['Ajuste_Rotten'],q=[0,0.95,1],\
                                         labels=["menores","maiores"]))
```

```python
melhores_series=melhores_series[(melhores_series['Categ_IMDB']=="maiores")&\
(melhores_series['Categ_Rotten']=="maiores")].sort_values(["Ajuste_Rotten","Ajuste_IMDB"]\
                                                          ,ascending=False)
```

```python
#%% Filmes
# Identificar os que estão com melhores notas nas duas avaliações
# Usar o percentil 95 das notas como referência
melhores_series[["Ajuste_IMDB","Ajuste_Rotten"]].quantile(0.95)
```

```python
# Gerando dados Melhores Filmes
melhores_filmes = melhores_filmes.assign(Categ_IMDB=pd.qcut(melhores_filmes['Ajuste_IMDB'],\
                                                            q=[0,0.95,1],labels=["maiores","menores"]))
```

```python
melhores_filmes = melhores_filmes.assign\
(Categ_Rotten = pd.qcut(melhores_filmes['Ajuste_Rotten'],q=[0,0.95,1],labels=["maiores","menores"]))
```
