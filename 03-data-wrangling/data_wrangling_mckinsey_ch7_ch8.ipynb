```python
# Handling missing data
# Pandas uses NaN (Not a Number) to represent missing data (for data with float64 dtype)
# The float64 data type is a 64-bit, double-precision floating-point format used to store decimal numbers in computing
# It is the standard numeric type for decimals in Python (and the default in libraries like NumPy and Pandas)

float_data = pd.Series([1.2,-3.5,np.nan,0])
float_data
```

```python
# method isna() gives us a Boolean Series where values are null
float_data.isna()
```

```python
# Built-in Python None is treated as NA (Not available)
string_data = pd.Series(["aardvark",np.nan,None,"avocado"])
string_data
```

```python
string_data.isna()
```

```python
float_data = pd.Series([1,2,None],dtype = "float64")
float_data
```

```python
# None é um objeto nativo do Python. Significa "ausência de valor" na linguagem em si. 
# Funciona bem em listas, dicionários, lógica condicional. O problema é que None é um objeto
# Num array de inteiros, inserir None força o pandas a converter tudo para object dtype 
# — porque agora o array precisa guardar tanto números quanto um objeto Python. Isso é lento e desperdiça memória.

# np.nan (Not a Number) vem do NumPy e é um valor especial do padrão IEEE 754 de ponto flutuante. É um float — não um objeto Python arbitrário.
# Isso significa que ele cabe dentro de um array numérico sem forçar conversão de tipo. Operações matemáticas com ele não geram erro, propagam o nan

# O que é float64
# É um tipo de dado numérico de ponto flutuante com 64 bits de precisão.
# O "64" refere-se à quantidade de bits usados para armazenar o número na memória — quanto mais bits, maior a precisão e o intervalo representável.
# É o tipo numérico padrão do NumPy e do pandas para dados decimais.
# Quando você cria uma Series com números e não especifica dtype, o pandas provavelmente vai usar float64.
```

```python
# NA handling object methods
# dropna - filter axis labels based on wheter valure for each label have missinf data
# fillna = fill in missing data with some value or usin an interpolation method such as "ffill" or "bfill"
# isna - return Boolean value indicatin which values are missing/NA
# notna - negation od isna
```

```python
data = pd.Series([1, np.nan, 3.5, np.nan, 7])
data
```

```python
data.dropna()
```

```python
# dropna equivale ao seguinte filtro:
data[data.notna()] #retorna somente as linhas com boleano True
```

```python
# Em um dataframe, dropna() remove cada linha que possua valor NA. se usar o argumento how = "all", será retirada somente a linha que tuver todos os valores nulos
data = pd.DataFrame([[1.,6.5,3.],[1,np.nan,np.nan],[np.nan,np.nan,np.nan],[np.nan,6.5,3.]])
data
```

```python
data.dropna()
```

```python
data.dropna(how="all")
```

```python
# para excluir colunas, inclua o argumento  axis = "columns"

data[4]=np.nan
data
```

```python
data.dropna(axis="columns",how="all")
```

```python
# argumento thresh
# thresh=x significa: mantenha a linha se ela tiver pelo menos x valores não-nulos

#criar o dataframe
df = pd.DataFrame(np.random.standard_normal((7,3)))
df.iloc[:4,1]=np.nan
df.iloc[:2,2]=np.nan
df
```

```python
df.dropna(thresh=2)
```

```python
## Prenchendo data faltante
```

```python
# função fillna
# permite preencher NA com um valor específico
# diminui o risco de descartar outros dados quando se descarta NA
df
```

```python
df.fillna(0)
```

```python
# Se usar um dicionário, é possível personalizar cada coluna com um valor diferente
df
```

```python
df.fillna({1:0.57,2:0.12})
```

```python
# função ffill - forward fill
# Propaga o último valor válido para baixo, preenchendo os NaNs seguintes com o valor anterior mais próximo
# O nome vem de forward — ele olha para trás (o valor que já passou) e preenche para frente
# parâmetro limit: limit=n restringe a propagação a no máximo n posições consecutivas a partir do último valor válido. O que exceder permanece NaN

# O inverso do ffill é o bfill (backward fill) — preenche com o próximo valor válido, olhando para frente.

df
```

```python
df.fillna(method="bfill")
```

```python
df.bfill()
```

```python
# pode-se preencher com cálculos estatísticos. e.g. média
df.fillna(df.mean())
```

```python
# pode-se preencher com cálculos estatísticos. e.g. mediana
df.fillna(df.median())
```
