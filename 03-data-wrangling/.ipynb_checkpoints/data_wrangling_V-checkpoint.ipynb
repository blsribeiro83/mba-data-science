```python
# IMPORTAÇÃO DE BANCOS DE DADOS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
# pip install xlrd # pacote dependência que read_excel usa para ler arquivos na extensão .xls
```

```python
#%% Importando o banco de dados

dados_jogos = pd.read_csv('(4.2) Jogos Copa 22.csv', sep=',')
```

```python
dados_jogos.info()
```

```python
dados_jogos["venceu"] = np.where((dados_jogos['team_home_score']-dados_jogos['team_away_score']>0),\
                                 "Mandante",np.where((dados_jogos['team_home_score']-dados_jogos['team_away_score']<0),\
                                                     "Visitante",np.where((dados_jogos['team_home_score']-dados_jogos['team_away_score']==0),\
                                                                          "Empate","Demais")))
```

```python
dados_jogos
```

```python
# TSubstituir a informação de penaltis para NaN e alterar para tipo float
dados_jogos["pens_home_score"] = dados_jogos['pens_home_score'].replace("False",np.nan)
dados_jogos['pens_away_score']=dados_jogos['pens_away_score'].replace("False",np.nan)
dados_jogos['pens_away_score']=dados_jogos['pens_away_score'].astype("float")
dados_jogos['pens_home_score']=dados_jogos['pens_home_score'].astype("float")
```

```python
dados_jogos
```

```python
# Ajustando a variável com o time vencedor
dados_jogos["vencedor"] = np.where(dados_jogos['pens_home_score']-dados_jogos['pens_home_score']>0,"Mandante",\
                                   np.where(dados_jogos['pens_home_score']-dados_jogos['pens_home_score']<0,\
                                            "Visitante",np.where(dados_jogos['pens_home_score'].isna(),dados_jogos["venceu"],\
                                            "Demais")))
#na última linha, onde não houve pênalti, usa o resultado do DataFrame anterior, com a variável "venceu"
```

```python
dados_jogos
```

```python
# eliminar a variável da base anterior
dados_jogos.drop(columns=["venceu"],inplace=True)
```

```python
dados_jogos
```

```python
# Tabela de frequências
dados_jogos["vencedor"].value_counts()
```

```python
# Histograma
plt.figure(figsize=(15,9),dpi=600)
ax = sns.countplot(x=dados_jogos["vencedor"], color="olive")
for container in ax.containers: ax.bar_label(container,padding=1,fontsize=12)
plt.xlabel("Vencedor",fontsize=15)
plt.ylabel("Contagem",fontsize=15)
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
plt.show()
```

```python
# Identificando a fase da competição
# Split divide os regisatros da variável conforme o divisor (neste caso ' '), e a segunda str pega o elemento [0]
dados_jogos["fase"] = dados_jogos['stage'].str.split(' ').str[0]
```

```python
dados_jogos["fase"].value_counts()
```

```python
# Histograma
plt.figure(figsize=(15,9),dpi=600)
ax1 = sns.countplot(data=dados_jogos,y="fase",hue="vencedor")
for container in ax1.containers: ax1.bar_label(container,padding=1,fontsize=12)
plt.legend(loc="upper right",title="Vencedor")
plt.xlabel("Quantidade",fontsize=14)
plt.ylabel("Fase",fontsize=14)
plt.xticks(np.arange(0,21,step=1))
plt.show()
```

```python
# Analisando se as informações de probabilidade estavam certas
# Categoriar a probabilidade
dados_jogos["previsao"]=np.where(dados_jogos['prediction_team_home_win']>dados_jogos['prediction_team_away_win'],\
                                 "Mandante",np.where(dados_jogos['prediction_team_home_win']<dados_jogos['prediction_team_away_win'],\
                                                     "Visitante","Demais"))
```

```python
# Comparar com o realizado
dados_jogos["resultado_previsao"]=np.where(dados_jogos['vencedor']==dados_jogos['previsao'],"Acerto",\
np.where(dados_jogos['vencedor']!=dados_jogos['previsao'],"Erro","Demais"))
```

```python
dados_jogos["resultado_previsao"].value_counts(normalize=True)
```

```python
# Analisando jogadores que fizeram gol
# Informação esta´na variável "events_list", mais complexa
# A informação está em action_player_1
# Abaixo, utilizamos regex (regular expression), úteis para trabalhar em strings
# É importante conhecer a existência delas, em alguns casos podem ajudar
# O str.extractall pede para extrair em todas as ocorrências do padrão
extrai_gol = dados_jogos['events_list'].str.extractall("'Goal', 'action_player_1': '\\w*(.*?)\\w*\\'")
```

```python
# Extraindo jogadores que fizeram pênaltis
extrai_penalti = dados_jogos['events_list'].str.extractall("'event_type':'Penalty','action_player_1':'\\w*(.*?)\\w*\\'")
```

```python
extrai_penalti
```

```python
# Unindo os datasets
gols = pd.concat([extrai_gol,extrai_penalti],ignore_index=True)
```

```python
tabela_artil=gols.value_counts()
tabela_artil
```
